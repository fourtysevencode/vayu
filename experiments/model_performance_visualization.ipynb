```python
import pandas as pd
from matplotlib import pyplot as plt
```

```python
df1 = pd.read_csv(r"C:\Users\HP\OneDrive\Documents\projects\vayu\model\vayu_v1\results.csv")
df2 = pd.read_csv(r"C:\Users\HP\OneDrive\Documents\projects\vayu\model\vayu_v2\results.csv")
df = pd.concat([df1,df2], ignore_index=True)
print(df)
```

```python
df.head()
```

```python
plt.figure(figsize=(10,3))
x = df.epoch
plt.xticks(list(x))
plt.plot(x, df["metrics/mAP50(B)"], "red", label="mAP50")
plt.plot(x, df["train/box_loss"], "blue", label="box loss")
plt.plot(x, df["metrics/recall(B)"], "green", label="recall")
plt.xlabel("Epoch")
plt.title("Variation of Metrics over Epochs in Training")
plt.legend()
```
